# Clasificación de celdas Voronoi (K=70): haversine + data augmentation + **SAM** (Sharpness-Aware Minimization)

Mismo experimento que `cell_classifier_k70_haversine_aug.ipynb` (mismo train balanceado y validación global,
ResNet18 preentrenada en ImageNet, dropout, **mismo data augmentation**, misma loss haversine, misma estrategia de finetuning,
misma selección de modelo y mismas métricas). **Lo único que cambia es el optimizador**: AdamW se envuelve con
**SAM** (Foret et al., 2021).

SAM busca pesos que estén en un **mínimo plano**: no alcanza con que la loss sea baja en $w$, tiene que seguir
siendo baja en todo un entorno de radio $\rho$ alrededor de $w$,

$$\min_w \; \max_{\lVert \epsilon \rVert_2 \le \rho} \; \mathcal{L}(w + \epsilon).$$

En cada batch se hacen **dos pasadas**:

1. Gradiente $g = \nabla_w \mathcal{L}(w)$ y perturbación adversaria **en los pesos**
   $\hat\epsilon = \rho\, g / \lVert g \rVert_2$ (el punto cercano donde la loss más sube).
2. Gradiente en $w + \hat\epsilon$; se vuelve a $w$ y AdamW da el paso con ese gradiente.

- Es la versión "en los pesos" del entrenamiento adversario de la Unidad 3: en vez de perturbar la imagen, se
  perturban los parámetros. Conecta con "ruido en los pesos favorece mínimos planos, más robustos".
- Los mínimos planos suelen generalizar mejor: el objetivo es achicar la brecha train–validación.
- `RHO_SAM` = 0.05 (valor estándar del paper). Costo: ~2× tiempo por época de train (validación igual).
- En la segunda pasada no se actualizan las estadísticas acumuladas de BatchNorm (la imagen es la misma; solo
  cambian los pesos), como en la implementación de referencia.
- Se reporta además la **nitidez** en train: $\mathcal{L}(w + \hat\epsilon) - \mathcal{L}(w)$, cuánto sube la
  loss en la dirección adversaria. Un mínimo más plano tiene nitidez más baja.

**Selección de modelo:** igual que la corrida sin SAM (loss haversine de validación en $w$, sin perturbar).

Orden: **1. Configuración y datos → 2. Entrenamiento → 3. Métricas sobre validación → 4. Métricas adicionales →
5. Comparación con la corrida sin SAM**.

### Carga, etiquetas y controles

Las imágenes se decodifican al solicitarlas. Controles:

- El CSV de prototipos tiene `ind_cluster` = 0..K-1 y su SHA-256 coincide con el registrado en el config del train
  (el train se muestreó con estas mismas celdas).
- La etiqueta de cada imagen (train y validación) se calcula desde latitud/longitud con el prototipo más cercano.
  En train se compara con la columna `ind_cluster` guardada por el muestreo: deben ser idénticas.
- No hay imágenes ni panoramas compartidos entre splits y todas las celdas tienen ejemplos de train.

No se eliminan filas.

```python
import geopandas as gpd

datos = {
    split: HFDataset.from_parquet(str(archivo), cache_dir=str(raiz / ".cache/modeling/hf_celdas"))
        .cast_column("image", HFImage(decode=False))
    for split, archivo in archivos.items()
}
prototipos = pd.read_csv(ruta_prototipos).sort_values(COL_CLUSTER).reset_index(drop=True)
NUM_CLASES = len(prototipos)
assert (prototipos[COL_CLUSTER].to_numpy() == np.arange(NUM_CLASES)).all()
coords_prototipos = prototipos[["lat", "lon"]].to_numpy(dtype=np.float64)

# El train tiene que haberse muestreado con estos mismos prototipos.
huella_prototipos = hashlib.sha256(ruta_prototipos.read_bytes()).hexdigest()
config_train = json.loads(ruta_config_train.read_text(encoding="utf-8"))
if config_train["prototipos"]["sha256"] != huella_prototipos:
    raise ValueError("El train se muestreó con otros prototipos: "
                     f"{config_train['prototipos']['ruta']} (sha256 {config_train['prototipos']['sha256'][:12]}...).")


def to_xyz(lat, lon):
    lat, lon = np.radians(lat), np.radians(lon)
    return np.column_stack([np.cos(lat) * np.cos(lon), np.cos(lat) * np.sin(lon), np.sin(lat)])


centroides = to_xyz(coords_prototipos[:, 0], coords_prototipos[:, 1])


def asignar_celda(lat, lon):
    # Celda del prototipo más cercano: mayor producto punto = menor distancia angular (igual que el muestreo).
    return (to_xyz(lat, lon) @ centroides.T).argmax(axis=1).astype(np.int64)

RUTA_PAISES = "src/notebooks/eda/assets/ne_110m_admin_0_countries.geojson"  # Natural Earth, ISO_A2_EH

def completar_country_code(tabla):
    # Completa country_code vacío o nulo con el país cuyo polígono contiene las coordenadas.
    # El mapa es de escala 1:110m: si un punto cae fuera de todo polígono (costa, islas), se usa el más cercano.
    vacios = tabla["country_code"].isna() | (tabla["country_code"].astype("string").str.strip() == "")
    if not vacios.any():
        return tabla
    lat_col, lon_col = COLS_COORDENADAS
    con_coords = vacios & tabla[[lat_col, lon_col]].notna().all(axis=1)

    paises = gpd.read_file(resolver_ruta(RUTA_PAISES)).to_crs("EPSG:4326")
    paises = paises.loc[paises["ISO_A2_EH"] != "-99", ["ISO_A2_EH", "geometry"]]
    puntos = gpd.GeoDataFrame(
        {"fila": tabla.index[con_coords]},
        geometry=gpd.points_from_xy(tabla.loc[con_coords, lon_col], tabla.loc[con_coords, lat_col]),
        crs="EPSG:4326",
    )
    dentro = puntos.sjoin(paises, how="left", predicate="within").drop_duplicates("fila")
    codigos = dentro.set_index("fila")["ISO_A2_EH"]
    sin_pais = codigos.index[codigos.isna()]
    if len(sin_pais):
        # La distancia se mide en un CRS proyectado; alcanza para elegir el polígono más cercano.
        cercanos = (puntos[puntos["fila"].isin(sin_pais)].to_crs("EPSG:3857")
                    .sjoin_nearest(paises.to_crs("EPSG:3857"), how="left").drop_duplicates("fila"))
        codigos.loc[cercanos["fila"].to_numpy()] = cercanos["ISO_A2_EH"].to_numpy()
    tabla = tabla.copy()
    tabla["country_code"] = tabla["country_code"].astype("object")
    tabla.loc[codigos.index, "country_code"] = codigos.to_numpy()
    print(f"country_code completado por coordenadas: {len(codigos):,} de {int(vacios.sum()):,} vacíos")
    return tabla

def metadatos(dataset):
    # La etiqueta no se lee del parquet: se calcula desde las coordenadas.
    tabla = dataset.select_columns(["country_code", "panoid", "image_id", *COLS_COORDENADAS]).to_pandas()
    tabla = completar_country_code(tabla)
    if tabla.isna().any().any():
        raise ValueError("Hay valores faltantes en los metadatos.")
    tabla["country_code"] = tabla["country_code"].str.strip().str.upper()
    lat, lon = (tabla[c].to_numpy(dtype=np.float64) for c in COLS_COORDENADAS)
    if not (np.isfinite(lat) & np.isfinite(lon) & (np.abs(lat) <= 90) & (np.abs(lon) <= 180)).all():
        raise ValueError("Hay coordenadas inválidas.")
    tabla[COL_CLUSTER] = asignar_celda(lat, lon)
    return tabla

meta_train, meta_val = (metadatos(datos[s]) for s in ("train", "validation"))

# Control: la columna guardada por el muestreo tiene que coincidir con la etiqueta calculada.
if COL_CLUSTER in datos["train"].column_names:
    guardada = np.asarray(datos["train"][COL_CLUSTER], dtype=np.int64)
    if not np.array_equal(guardada, meta_train[COL_CLUSTER].to_numpy()):
        raise ValueError(f"{COL_CLUSTER} guardado en train no coincide con el calculado "
                         f"({(guardada != meta_train[COL_CLUSTER].to_numpy()).sum()} filas).")
    print("Control OK: la etiqueta calculada coincide con la columna del train.")
if not len(meta_train) or not len(meta_val):
    raise ValueError("Ambos splits deben contener imágenes.")
for columna in ("panoid", "image_id"):
    if set(meta_train[columna]) & set(meta_val[columna]):
        raise ValueError(f"Fuga de datos: {columna} compartidos entre splits.")
conteo = pd.DataFrame({
    "train": meta_train[COL_CLUSTER].value_counts(),
    "validation": meta_val[COL_CLUSTER].value_counts(),
}).reindex(range(NUM_CLASES)).fillna(0).astype(int)
if (conteo["train"] == 0).any():
    raise ValueError(f"Celdas sin ejemplos de train: {conteo.index[conteo['train'] == 0].tolist()}")
print(f"Clases: {NUM_CLASES}; train: {len(meta_train):,}; validación: {len(meta_val):,}")
print(f"Imágenes por celda (train): min {conteo['train'].min()}, mediana {int(conteo['train'].median())}, máx {conteo['train'].max()}")
celdas_sin_val = conteo.index[conteo["validation"] == 0].tolist()
print(f"Celdas sin imágenes de validación: {celdas_sin_val} (sin métricas por celda; no entran en promedios macro)")
conteo.T
```

```python
# Media y desvío de ImageNet: los pesos preentrenados esperan esta normalización.
PESOS = models.ResNet18_Weights.IMAGENET1K_V1
MEDIA_IMAGENET, STD_IMAGENET = PESOS.transforms().mean, PESOS.transforms().std

# Augmentation suave y uniforme, solo en train. Sin flip (lado de manejo, textos), sin hue (el color de la
# tierra y la vegetación es pista geográfica) y sin borrado de parches (puede tapar la pista clave).
transform_train = transforms.Compose([
    transforms.RandomResizedCrop(TAMANO_IMAGEN, scale=(0.6, 1.0), ratio=(0.9, 1.1)),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0),
    transforms.ToTensor(),
    transforms.Normalize(MEDIA_IMAGENET, STD_IMAGENET),
])
# Validación: preprocesamiento determinístico, igual que en los demás clasificadores.
transform_val = transforms.Compose([
    transforms.Resize(TAMANO_IMAGEN),
    transforms.CenterCrop(TAMANO_IMAGEN),
    transforms.ToTensor(),
    transforms.Normalize(MEDIA_IMAGENET, STD_IMAGENET),
])

def haversine_km(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.deg2rad, (lat1, lon1, lat2, lon2))
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 2 * RADIO_TIERRA_KM * np.arcsin(np.sqrt(np.clip(a, 0, 1)))


def distancias_a_prototipos(metadatos):
    # Matriz (imágenes x celdas) con la distancia haversine en km de cada foto a cada prototipo.
    lat, lon = (metadatos[c].to_numpy(dtype=np.float64)[:, None] for c in COLS_COORDENADAS)
    return haversine_km(lat, lon, coords_prototipos[None, :, 0], coords_prototipos[None, :, 1])


class ImagenesCeldas(Dataset):
    # Devuelve (imagen, celda, distancias en km a los K prototipos). Las distancias arman el target blando.
    def __init__(self, datos, metadatos, transform):
        self.datos = datos
        self.etiquetas = metadatos[COL_CLUSTER].to_numpy(dtype=np.int64)
        distancias = distancias_a_prototipos(metadatos)
        # Coherencia: el prototipo más cercano en km es la celda asignada por distancia angular.
        if not np.array_equal(distancias.argmin(axis=1), self.etiquetas):
            raise ValueError("La celda más cercana en km no coincide con la etiqueta.")
        self.distancias = torch.from_numpy(distancias).float()
        self.transform = transform

    def __len__(self):
        return len(self.datos)

    def __getitem__(self, idx):
        contenido = self.datos[int(idx)]["image"]
        if contenido.get("bytes") is not None:
            fuente = BytesIO(contenido["bytes"])
        else:
            fuente = resolver_ruta(contenido["path"])
        with Image.open(fuente) as imagen:
            tensor = self.transform(imagen.convert("RGB"))
        return tensor, int(self.etiquetas[idx]), self.distancias[idx]

dataset_train = ImagenesCeldas(datos["train"], meta_train, transform_train)
dataset_val = ImagenesCeldas(datos["validation"], meta_val, transform_val)
loader_train = DataLoader(dataset_train, batch_size=BATCH_SIZE, shuffle=True,
                          generator=torch.Generator().manual_seed(SEMILLA),
                          num_workers=NUM_WORKERS, pin_memory=device.type == "cuda")
loader_val = DataLoader(dataset_val, batch_size=BATCH_SIZE, shuffle=False,
                        num_workers=NUM_WORKERS, pin_memory=device.type == "cuda")
```

### Elección de τ

Para cada τ, cuánta masa del target blando queda en la celda verdadera (media y percentil 10), en las 3 celdas más
cercanas y en celdas a más de 2000 km. Un τ razonable deja la mayor parte en la celda verdadera y casi nada lejos;
con el τ elegido el target es parecido a un label smoothing fuerte, pero concentrado en las celdas vecinas.

```python
d_train = dataset_train.distancias.double()
etiquetas_train = torch.from_numpy(meta_train[COL_CLUSTER].to_numpy(dtype=np.int64))
filas = []
for tau in sorted({100, 200, 300, 400, 500, 750, 1000, TAU_KM}):
    q = torch.softmax(-d_train / tau, dim=1)
    masa = q.gather(1, etiquetas_train[:, None]).squeeze(1).numpy()
    filas.append({"tau_km": tau, "masa_celda_real_media": masa.mean(), "masa_celda_real_p10": np.percentile(masa, 10),
                  "masa_top3": q.topk(3, dim=1).values.sum(dim=1).mean().item(),
                  "masa_a_mas_de_2000_km": (q * (d_train > 2000)).sum(dim=1).mean().item()})
tabla_tau = pd.DataFrame(filas).set_index("tau_km").round(3)
display(tabla_tau.style.apply(lambda f: ["font-weight: bold" if f.name == TAU_KM else ""] * len(f), axis=1))
```

## 2. Fase de entrenamiento

Finetuning de ResNet18 (`IMAGENET1K_V1`) con la capa final reemplazada por `Dropout(DROPOUT)` + una lineal de
`NUM_CLASES` salidas (logits),
CrossEntropy contra el target haversine (`TAU_KM`) y **SAM sobre AdamW** (`RHO_SAM`) con dos learning rates: `LR_CABEZA` para la capa nueva y
`LR_BACKBONE` para el resto. Durante las primeras `EPOCAS_CABEZA` épocas el backbone queda congelado y con
BatchNorm en modo eval; esas épocas no mueven el scheduler ni el contador de early stopping. Se reportan loss,
accuracy top-1 y balanced accuracy por época. Se guardan `mejor_resnet18_celdas.pt` (menor loss de validación),
`ultimo_resnet18_celdas.pt` (última época) e `historial.csv`.

Las métricas de train (loss, accuracy, error en km) se calculan sobre las imágenes **aumentadas**, así que
la brecha entre train y validación no se compara uno a uno con la corrida sin augmentation. Para comparar
los dos modelos, mirar las métricas de validación.

En el historial, `loss_perturbada_train` es la loss en los pesos perturbados $w + \hat\epsilon$ (segunda pasada); la nitidez es su diferencia con `loss_train`.

```python
def construir_modelo(preentrenado=True):
    # Los pesos de ImageNet traen 1000 clases: se carga la red completa y se reemplaza la capa final.
    m = models.resnet18(weights=PESOS if preentrenado else None)
    m.fc = nn.Sequential(nn.Dropout(DROPOUT), nn.Linear(m.fc.in_features, NUM_CLASES))
    return m

modelo = construir_modelo().to(device)
backbone_congelado = False

def fijar_backbone(congelado):
    # Congelado: sin gradientes en el backbone (y BatchNorm en eval durante el entrenamiento, ver ejecutar_epoca).
    global backbone_congelado
    backbone_congelado = congelado
    for nombre, p in modelo.named_parameters():
        if not nombre.startswith("fc."):
            p.requires_grad = not congelado

criterio = nn.CrossEntropyLoss()  # CE dura: métrica comparable con el modelo de CrossEntropy.


def objetivo_haversine(distancias, tau=TAU_KM):
    # Target blando: softmax(-d / τ) sobre las distancias (km) de la foto a cada prototipo.
    return torch.softmax(-distancias / tau, dim=1)


def criterio_haversine(logits, distancias):
    # CrossEntropy contra el target blando (haversine label smoothing).
    return -(objetivo_haversine(distancias) * torch.log_softmax(logits, dim=1)).sum(dim=1).mean()
optimizador = torch.optim.AdamW([
    {"params": modelo.fc.parameters(), "lr": LR_CABEZA},
    {"params": [p for n, p in modelo.named_parameters() if not n.startswith("fc.")], "lr": LR_BACKBONE},
], weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizador, factor=0.1, patience=2)


class SAM:
    # Sharpness-Aware Minimization (Foret et al., 2021) sobre un optimizador base (acá AdamW).
    # El scheduler y el weight decay siguen actuando sobre el optimizador base.
    def __init__(self, optimizador_base, rho):
        self.base, self.rho, self.perturbaciones = optimizador_base, rho, {}

    def _con_gradiente(self):
        return [p for g in self.base.param_groups for p in g["params"] if p.grad is not None]

    @torch.no_grad()
    def primer_paso(self):
        # w -> w + e, con e = rho * g / ||g|| (norma L2 global sobre todos los parámetros entrenables).
        params = self._con_gradiente()
        norma = torch.norm(torch.stack([p.grad.norm(2) for p in params]), 2)
        escala = self.rho / (norma + 1e-12)
        for p in params:
            e = p.grad * escala
            p.add_(e)
            self.perturbaciones[p] = e
        self.base.zero_grad(set_to_none=True)

    @torch.no_grad()
    def segundo_paso(self):
        # Vuelve a w y da el paso del optimizador base con el gradiente calculado en w + e.
        for p, e in self.perturbaciones.items():
            p.sub_(e)
        self.perturbaciones = {}
        self.base.step()
        self.base.zero_grad(set_to_none=True)


@contextmanager
def bn_sin_actualizar(red):
    # Durante la pasada con pesos perturbados no se tocan las estadísticas acumuladas de BatchNorm
    # (momentum 0: running = running). El batch se sigue normalizando con su propia media y varianza.
    bns = [m for m in red.modules() if isinstance(m, nn.modules.batchnorm._BatchNorm)]
    momentos = [m.momentum for m in bns]
    for m in bns:
        m.momentum = 0.0
    try:
        yield
    finally:
        for m, momento in zip(bns, momentos):
            m.momentum = momento


sam = SAM(optimizador, RHO_SAM)
scaler = torch.amp.GradScaler("cuda", enabled=amp_activo)
print(f"Parámetros totales: {sum(p.numel() for p in modelo.parameters()):,}; "
      f"capa final: {sum(p.numel() for p in modelo.fc.parameters()):,}")

def ejecutar_epoca(modelo, loader, entrenar):
    # Devuelve loss haversine (la que se optimiza, en w), CE dura, accuracy, balanced accuracy, error mediano en
    # km y, en train, la loss en los pesos perturbados de SAM.
    modelo.train(entrenar)
    if entrenar and backbone_congelado:
        for nombre, modulo in modelo.named_children():
            if nombre != "fc":
                modulo.eval()  # Conserva las estadísticas de BatchNorm de ImageNet.
    total_loss, total_perturbada, total_ce, aciertos, total = 0.0, 0.0, 0.0, 0, 0
    todas_pred, todas_real, errores_km = [], [], []
    if entrenar:
        optimizador.zero_grad(set_to_none=True)
    with torch.set_grad_enabled(entrenar):
        for paso, (imagenes, etiquetas, distancias) in enumerate(tqdm(loader, desc="Train" if entrenar else "Validación")):
            imagenes = imagenes.to(device, non_blocking=True)
            etiquetas = etiquetas.to(device, non_blocking=True)
            distancias = distancias.to(device, non_blocking=True)
            with torch.autocast(device_type=device.type, enabled=amp_activo):
                logits = modelo(imagenes)
            logits = logits.float()
            loss = criterio_haversine(logits, distancias)
            n = etiquetas.size(0)
            if entrenar:
                # SAM: 1) gradiente en w y salto a w + e; 2) gradiente en w + e, vuelta a w y paso de AdamW.
                loss.backward()
                sam.primer_paso()
                with bn_sin_actualizar(modelo):
                    loss_perturbada = criterio_haversine(modelo(imagenes).float(), distancias)
                loss_perturbada.backward()
                sam.segundo_paso()
                total_perturbada += loss_perturbada.detach().item() * n
            logits = logits.detach()
            pred_batch = logits.argmax(dim=1)
            total_loss += loss.detach().item() * n
            total_ce += criterio(logits, etiquetas).item() * n
            aciertos += (pred_batch == etiquetas).sum().item()
            total += n
            todas_pred.append(pred_batch.cpu())
            todas_real.append(etiquetas.cpu())
            errores_km.append(distancias.gather(1, pred_batch[:, None]).squeeze(1).cpu())
    with warnings.catch_warnings():  # Celdas predichas sin imágenes reales (p. ej. sin validación).
        warnings.simplefilter("ignore", UserWarning)
        acc_balanceada = balanced_accuracy_score(torch.cat(todas_real).numpy(), torch.cat(todas_pred).numpy())
    resultado = {"loss": total_loss / total, "ce": total_ce / total, "accuracy": aciertos / total,
                 "balanced_accuracy": acc_balanceada, "error_mediano_km": float(torch.cat(errores_km).median())}
    if entrenar:
        resultado["loss_perturbada"] = total_perturbada / total
    return resultado
```

```python
ruta_resultados.mkdir(parents=True, exist_ok=True)
ruta_checkpoint = ruta_resultados / "mejor_resnet18_celdas.pt"
ruta_ultimo = ruta_resultados / "ultimo_resnet18_celdas.pt"
if ruta_checkpoint.exists():
    raise FileExistsError("Ya existe un modelo en esta carpeta. Usá otra RUTA_RESULTADOS para entrenar.")
config_entrenamiento = {
    "arquitectura": "resnet18", "tarea": "clasificacion_celdas_voronoi", "weights": "IMAGENET1K_V1",
    "modo": "finetuning", "epocas_cabeza": EPOCAS_CABEZA,
    "ruta_train": str(archivos["train"]), "ruta_validation": str(archivos["validation"]),
    "prototipos": {"ruta": str(ruta_prototipos), "sha256": huella_prototipos},
    "etiqueta": "calculada al vuelo: prototipo más cercano por distancia angular",
    "num_clases": NUM_CLASES, "columna_etiqueta": COL_CLUSTER, "celdas_sin_validacion": celdas_sin_val,
    "epocas_maximas": EPOCAS, "batch_size": BATCH_SIZE, "acumular_batches": ACUMULAR_BATCHES,
    "lr_cabeza": LR_CABEZA, "lr_backbone": LR_BACKBONE, "optimizador": "SAM(AdamW)", "rho_sam": RHO_SAM, "weight_decay": WEIGHT_DECAY,
    "loss": "CrossEntropy con haversine label smoothing", "tau_km": TAU_KM, "dropout": DROPOUT,
    "criterio_seleccion": "loss haversine de validación",
    "data_augmentation": "RandomResizedCrop(scale=0.6-1, ratio=0.9-1.1) + ColorJitter(0.2, sin hue)",
    "paciencia": PACIENCIA, "tamano_imagen": TAMANO_IMAGEN, "semilla": SEMILLA,
    "amp": amp_activo, "pais_a_continente": pais_a_continente,
    "preprocesamiento_train": repr(transform_train), "preprocesamiento_val": repr(transform_val),
    "filas_train": len(dataset_train), "filas_validation": len(dataset_val),
    "torch": str(torch.__version__), "dispositivo": str(device),
}
(ruta_resultados / "config_entrenamiento.json").write_text(
    json.dumps(config_entrenamiento, ensure_ascii=False, indent=2), encoding="utf-8")
prototipos.to_csv(ruta_resultados / "prototipos.csv", index=False)  # para usar el modelo sin el sample

def guardar(ruta, epoca, loss_val, acc_val):
    torch.save({"model_state_dict": modelo.state_dict(), "epoca": epoca, "loss_validation": loss_val,
                "accuracy_validation": acc_val, "config": config_entrenamiento}, ruta)

historial = []
mejor_loss, sin_mejora = float("inf"), 0
for epoca in range(1, EPOCAS + 1):
    inicio = time.perf_counter()
    solo_cabeza = epoca <= EPOCAS_CABEZA
    fijar_backbone(solo_cabeza)
    lr_cabeza, lr_backbone = (g["lr"] for g in optimizador.param_groups)
    tr = ejecutar_epoca(modelo, loader_train, entrenar=True)
    va = ejecutar_epoca(modelo, loader_val, entrenar=False)
    loss_train, acc_train = tr["loss"], tr["accuracy"]
    loss_val, acc_val, bal_val = va["loss"], va["accuracy"], va["balanced_accuracy"]
    if not np.isfinite([loss_train, loss_val]).all():
        raise RuntimeError("Loss no finita. Revisá los learning rates y los datos.")
    historial.append({"epoca": epoca, "fase": "cabeza" if solo_cabeza else "completo",
                      **{f"{k}_train": v for k, v in tr.items()},
                      **{f"{k}_validation": v for k, v in va.items()},
                      "lr_cabeza": lr_cabeza, "lr_backbone": lr_backbone, "segundos": time.perf_counter() - inicio})
    pd.DataFrame(historial).to_csv(ruta_resultados / "historial.csv", index=False)
    guardar(ruta_ultimo, epoca, loss_val, acc_val)
    if loss_val < mejor_loss:
        mejor_loss, sin_mejora = loss_val, 0
        guardar(ruta_checkpoint, epoca, loss_val, acc_val)
    elif not solo_cabeza:
        sin_mejora += 1
    if not solo_cabeza:
        scheduler.step(loss_val)
    print(f"Época {epoca}/{EPOCAS} [{'cabeza' if solo_cabeza else 'completo'}] "
          f"| train loss={loss_train:.4f} nitidez={tr['loss_perturbada'] - loss_train:.4f} acc={acc_train:.3f} "
          f"| val loss={loss_val:.4f} ce={va['ce']:.4f} acc={acc_val:.3f} bal={bal_val:.3f} "
          f"km_med={va['error_mediano_km']:.0f} "
          f"| lr cabeza={lr_cabeza:.5g} backbone={lr_backbone:.5g}")
    if sin_mejora >= PACIENCIA:
        print(f"Early stopping: {PACIENCIA} épocas sin mejora.")
        break
print(f"Mejor loss haversine de validación: {mejor_loss:.4f}; checkpoint: {ruta_checkpoint}")
```

## 3. Métricas sobre validación

Se recarga el mejor checkpoint y se evalúa todo el split de validación. Para evaluar una corrida anterior sin
entrenar, indicar su carpeta en `RUTA_RESULTADOS`, ejecutar la sección 1 y luego esta sección.

- **Accuracy top-1 y top-5** sobre `ind_cluster`.
- `loss_validation` es la **CE dura** (comparable con el modelo de CrossEntropy); `loss_haversine_validation`
  es la loss optimizada.
- **Predicción por mínimo riesgo** (`_mbr`): en vez del `argmax`, la celda $k$ que minimiza
  $\sum_j p_j\, d(c_j, c_k)$, el error esperado en km según el modelo. Cuando duda entre celdas lejanas,
  elige una intermedia.
- **Error en km** (Haversine) entre la ubicación real y el prototipo de la celda predicha; también el error
  que resultaría de predecir siempre la celda correcta (cota inferior del esquema de celdas).
- Resultados globales y por continente real, y métricas por celda.

Como validación también selecciona el checkpoint, estas métricas no reemplazan un test independiente.

```python
checkpoint = torch.load(ruta_resultados / "mejor_resnet18_celdas.pt", map_location="cpu", weights_only=True)
cfg = checkpoint["config"]
if (cfg["arquitectura"] != "resnet18" or cfg.get("tarea") != "clasificacion_celdas_voronoi"
        or cfg["num_clases"] != NUM_CLASES or cfg["tamano_imagen"] != TAMANO_IMAGEN
        or cfg["preprocesamiento_val"] != repr(transform_val)
        or cfg.get("prototipos", {}).get("sha256") != huella_prototipos
        or cfg.get("tau_km") != TAU_KM or cfg.get("rho_sam") != RHO_SAM):
    raise ValueError("La configuración de evaluación no coincide con el entrenamiento.")
if "optimizador" in globals():
    optimizador.state.clear()
if "modelo" not in globals():
    modelo = construir_modelo(preentrenado=False).to(device)  # Los pesos salen del checkpoint.
modelo.load_state_dict(checkpoint["model_state_dict"])
print(f"Evaluando época {checkpoint['epoca']} (loss val={checkpoint['loss_validation']:.4f}, acc={checkpoint['accuracy_validation']:.3f})")
del checkpoint
modelo.eval()
salidas = []
with torch.inference_mode():
    for imagenes, *_ in tqdm(loader_val, desc="Métricas de validación"):
        with torch.autocast(device_type=device.type, enabled=amp_activo):
            logits = modelo(imagenes.to(device, non_blocking=True))
        salidas.append(logits.float().cpu())
logits = torch.cat(salidas)
assert logits.shape == (len(dataset_val), NUM_CLASES) and len(meta_val) == len(dataset_val)
probs = torch.softmax(logits, dim=1)
top5 = probs.topk(min(5, NUM_CLASES), dim=1).indices.numpy()
pred = top5[:, 0]
real = meta_val[COL_CLUSTER].to_numpy()

def haversine_km(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.deg2rad, (lat1, lon1, lat2, lon2))
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 2 * RADIO_TIERRA_KM * np.arcsin(np.sqrt(np.clip(a, 0, 1)))

lat_real, lon_real = (meta_val[c].to_numpy(dtype=np.float64) for c in COLS_COORDENADAS)
predicciones = meta_val[["image_id", "panoid", "country_code", *COLS_COORDENADAS]].copy()
predicciones["cluster_real"] = real
predicciones["cluster_predicho"] = pred
predicciones["prob_predicha"] = probs.max(dim=1).values.numpy()
predicciones["acierto_top1"] = pred == real
predicciones["acierto_top5"] = (top5 == real[:, None]).any(axis=1)
predicciones["error_km"] = haversine_km(lat_real, lon_real, coords_prototipos[pred, 0], coords_prototipos[pred, 1])
predicciones["error_km_cota"] = haversine_km(lat_real, lon_real, coords_prototipos[real, 0], coords_prototipos[real, 1])
# Predicción por mínimo riesgo: celda que minimiza el error esperado en km bajo las probabilidades del modelo.
distancias_prototipos = haversine_km(coords_prototipos[:, None, 0], coords_prototipos[:, None, 1],
                                     coords_prototipos[None, :, 0], coords_prototipos[None, :, 1])
pred_mbr = (probs.numpy().astype(np.float64) @ distancias_prototipos).argmin(axis=1)
predicciones["cluster_predicho_mbr"] = pred_mbr
predicciones["acierto_top1_mbr"] = pred_mbr == real
predicciones["error_km_mbr"] = haversine_km(lat_real, lon_real, coords_prototipos[pred_mbr, 0], coords_prototipos[pred_mbr, 1])
predicciones["continente_real"] = predicciones["country_code"].map(pais_a_continente)

resumen_metricas = {
    "n_validation": len(predicciones),
    "loss_validation": float(criterio(logits, torch.from_numpy(real.astype(np.int64))).item()),  # CE dura
    "loss_haversine_validation": float(criterio_haversine(logits, dataset_val.distancias).item()),
    "accuracy_top1": float(predicciones["acierto_top1"].mean()),
    "accuracy_top5": float(predicciones["acierto_top5"].mean()),
    "error_medio_km": float(predicciones["error_km"].mean()),
    "error_mediano_km": float(predicciones["error_km"].median()),
    "error_medio_km_cota_celdas": float(predicciones["error_km_cota"].mean()),
    "accuracy_top1_mbr": float(predicciones["acierto_top1_mbr"].mean()),
    "error_medio_km_mbr": float(predicciones["error_km_mbr"].mean()),
    "error_mediano_km_mbr": float(predicciones["error_km_mbr"].median()),
}
display(pd.Series(resumen_metricas, name="Validación").to_frame())
por_continente = predicciones.groupby("continente_real").agg(
    n=("error_km", "size"), accuracy_top1=("acierto_top1", "mean"), accuracy_top5=("acierto_top5", "mean"),
    error_medio_km=("error_km", "mean"), error_mediano_km=("error_km", "median"),
    error_medio_km_mbr=("error_km_mbr", "mean"), error_mediano_km_mbr=("error_km_mbr", "median"))
display(por_continente)
por_celda = predicciones.groupby("cluster_real").agg(
    n=("acierto_top1", "size"), accuracy_top1=("acierto_top1", "mean"), error_medio_km=("error_km", "mean")
).reindex(range(NUM_CLASES))  # Celdas sin validación: NaN.
por_celda["n"] = por_celda["n"].fillna(0).astype(int)
por_celda.index.name = "cluster_real"
predicciones.to_csv(ruta_resultados / "predicciones_validation.csv", index=False)
por_continente.to_csv(ruta_resultados / "metricas_por_continente.csv")
por_celda.to_csv(ruta_resultados / "metricas_por_celda.csv")
(ruta_resultados / "metricas_validation.json").write_text(
    json.dumps(resumen_metricas, ensure_ascii=False, indent=2, allow_nan=False), encoding="utf-8")
```

## 4. Métricas adicionales de clasificación

Usan `real`, `pred`, `probs` y `predicciones` de la sección 3. Por el desbalance entre celdas, además de la
accuracy global se reportan **balanced accuracy** (media del recall por celda) y **F1 macro** (todas las celdas
pesan igual), junto con **F1 ponderado** (pesa por cantidad de imágenes).

Los promedios macro se calculan **solo sobre las celdas con imágenes de validación**: una celda sin ejemplos
no tiene recall definido y con `zero_division=0` contaría como 0, bajando el promedio de forma artificial.
En el reporte por celda, esas celdas quedan con recall y F1 en NaN.

```python
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, classification_report,
                             cohen_kappa_score, f1_score, precision_score, recall_score)

celdas_con_val = np.unique(real)  # Solo celdas con imágenes de validación.
with warnings.catch_warnings():
    warnings.simplefilter("ignore", UserWarning)
    bal_acc = balanced_accuracy_score(real, pred)  # Ya ignora las celdas sin ejemplos reales.
resumen_extra = {
    "accuracy_top1": accuracy_score(real, pred),
    "balanced_accuracy": bal_acc,
    "precision_macro": precision_score(real, pred, labels=celdas_con_val, average="macro", zero_division=0),
    "recall_macro": recall_score(real, pred, labels=celdas_con_val, average="macro", zero_division=0),
    "f1_macro": f1_score(real, pred, labels=celdas_con_val, average="macro", zero_division=0),
    "f1_ponderado": f1_score(real, pred, labels=celdas_con_val, average="weighted", zero_division=0),
    "celdas_evaluadas": len(celdas_con_val),
    "cohen_kappa": cohen_kappa_score(real, pred),
    "baseline_celda_mayoritaria": float(np.bincount(meta_train[COL_CLUSTER], minlength=NUM_CLASES).max() / len(meta_train)),
    "baseline_azar": 1 / NUM_CLASES,
}
display(pd.Series(resumen_extra, name="Validación").to_frame())

reporte = pd.DataFrame(classification_report(real, pred, labels=range(NUM_CLASES), output_dict=True, zero_division=0)).T
reporte_celdas = reporte.iloc[:NUM_CLASES].copy()
reporte_celdas.index = reporte_celdas.index.astype(int)
reporte_celdas.index.name = COL_CLUSTER
reporte_celdas["n_train"] = conteo["train"]
reporte_celdas.loc[reporte_celdas["support"] == 0, ["recall", "f1-score"]] = np.nan  # Sin validación.
# Precisión: si la celda nunca fue predicha, no está definida.
reporte_celdas.loc[np.bincount(pred, minlength=NUM_CLASES) == 0, "precision"] = np.nan
display(reporte_celdas.sort_values("f1-score", ascending=False).round(3))
reporte_celdas.to_csv(ruta_resultados / "reporte_clasificacion_por_celda.csv")
(ruta_resultados / "metricas_extra_validation.json").write_text(
    json.dumps({k: float(v) for k, v in resumen_extra.items()}, indent=2), encoding="utf-8")
```

### Matriz de confusión con conteos

La matriz de la sección 3 está normalizada por fila (recall por celda). Acá se muestra con **conteos absolutos**
(escala logarítmica) y, abajo, normalizada por **columna** (precisión: de lo que el modelo dice que es la celda
*j*, qué fracción lo es de verdad).

```python
conteos_mc = np.zeros((NUM_CLASES, NUM_CLASES), dtype=int)
np.add.at(conteos_mc, (real, pred), 1)
por_columna = conteos_mc / np.maximum(conteos_mc.sum(axis=0, keepdims=True), 1)
pd.DataFrame(conteos_mc).to_csv(ruta_resultados / "matriz_confusion_conteos.csv", index_label="real")

from matplotlib.colors import LogNorm
fig, axes = plt.subplots(1, 2, figsize=(17, 7))
im0 = axes[0].imshow(conteos_mc + 0.0, cmap="magma_r", norm=LogNorm(vmin=1, vmax=max(conteos_mc.max(), 2)))
axes[0].set(xlabel="Celda predicha", ylabel="Celda real", title="Conteos (escala log)")
fig.colorbar(im0, ax=axes[0])
im1 = axes[1].imshow(por_columna, cmap="viridis", vmin=0, vmax=1)
axes[1].set(xlabel="Celda predicha", ylabel="Celda real", title="Normalizada por columna (precisión)")
fig.colorbar(im1, ax=axes[1])
fig.tight_layout()
fig.savefig(ruta_resultados / "matriz_confusion_conteos.png", dpi=160)
plt.show()

# Confusiones más frecuentes fuera de la diagonal, con la distancia entre prototipos.
fuera = conteos_mc.copy()
np.fill_diagonal(fuera, 0)
r, c = np.unravel_index(np.argsort(fuera, axis=None)[::-1][:15], fuera.shape)
top_confusiones = pd.DataFrame({
    "celda_real": r, "celda_predicha": c, "n": fuera[r, c],
    "pct_de_la_celda_real": (fuera[r, c] / conteos_mc.sum(axis=1)[r]).round(3),
    "distancia_prototipos_km": haversine_km(coords_prototipos[r, 0], coords_prototipos[r, 1],
                                            coords_prototipos[c, 0], coords_prototipos[c, 1]).round(0),
})
display(top_confusiones[top_confusiones["n"] > 0])
top_confusiones.to_csv(ruta_resultados / "top_confusiones.csv", index=False)
```

### Calibración de la confianza

Compara la probabilidad que el modelo asigna a su predicción con la accuracy real. Una curva por debajo de la
diagonal indica sobreconfianza. ECE = diferencia media ponderada entre confianza y accuracy por bin.

```python
conf = predicciones["prob_predicha"].to_numpy()
acierto = predicciones["acierto_top1"].to_numpy()
bordes = np.linspace(0, 1, 11)
bin_id = np.clip(np.digitize(conf, bordes[1:-1]), 0, 9)
calib = pd.DataFrame({
    "conf_media": [conf[bin_id == b].mean() if (bin_id == b).any() else np.nan for b in range(10)],
    "accuracy": [acierto[bin_id == b].mean() if (bin_id == b).any() else np.nan for b in range(10)],
    "n": [(bin_id == b).sum() for b in range(10)],
})
ece = float(np.nansum(calib["n"] / calib["n"].sum() * (calib["accuracy"] - calib["conf_media"]).abs()))
print(f"ECE = {ece:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot([0, 1], [0, 1], "--", color="gray", label="Calibración perfecta")
axes[0].plot(calib["conf_media"], calib["accuracy"], marker="o", label="Modelo")
axes[0].set(xlabel="Confianza (prob. predicha)", ylabel="Accuracy", title=f"Curva de calibración (ECE={ece:.3f})",
            xlim=(0, 1), ylim=(0, 1))
axes[0].legend()
axes[1].hist([conf[acierto], conf[~acierto]], bins=20, range=(0, 1), label=["Aciertos", "Errores"], stacked=True)
axes[1].set(xlabel="Confianza", ylabel="Imágenes", title="Confianza en aciertos y errores")
axes[1].legend()
for ax in axes:
    ax.grid(alpha=0.3)
fig.tight_layout()
fig.savefig(ruta_resultados / "calibracion.png", dpi=160)
plt.show()
```

## 5. Comparación con la corrida sin SAM

Se leen las métricas y predicciones guardadas por `cell_classifier_k70_haversine_aug.ipynb` (misma
configuración, sin SAM, misma validación). Referencia adicional: el snapshot ensemble de ese modelo
(`..._aug_snapshot_ensemble.ipynb`) llegó a top-1 0.343 y error mediano 846 km, pero con 5 modelos.

La significancia se estima con un **bootstrap pareado por panorama** (las 4 vistas de un panorama no son
independientes): se remuestrean panoramas con reposición y se recalcula la diferencia SAM − sin SAM en accuracy
y en error mediano. Si el intervalo del 95% no contiene el 0, la diferencia es significativa.

```python
RUTA_BASELINE = "outputs/resnet18_IMAGENET_celdas_k70_haversine_aug/20261109_180324_467412"  # Sin SAM
B_BOOTSTRAP = 2000


def leer_metricas(ruta):
    return {**json.loads((ruta / "metricas_validation.json").read_text(encoding="utf-8")),
            **json.loads((ruta / "metricas_extra_validation.json").read_text(encoding="utf-8"))}


ruta_baseline = resolver_ruta(RUTA_BASELINE)
cfg_base = json.loads((ruta_baseline / "config_entrenamiento.json").read_text(encoding="utf-8"))
if cfg_base["preprocesamiento_train"] != repr(transform_train) or cfg_base.get("tau_km") != TAU_KM:
    raise ValueError("La corrida de referencia no tiene la misma configuración (augmentation o τ).")
esta_corrida = {**resumen_metricas, **{k: float(v) for k, v in resumen_extra.items()}}
filas_comparacion = ["accuracy_top1", "accuracy_top5", "balanced_accuracy", "f1_macro", "loss_validation",
                     "error_medio_km", "error_mediano_km", "accuracy_top1_mbr", "error_medio_km_mbr", "error_mediano_km_mbr"]
base_metricas = leer_metricas(ruta_baseline)
comparacion = pd.DataFrame({"sin SAM": {k: base_metricas[k] for k in filas_comparacion},
                            f"SAM (ρ={RHO_SAM})": {k: esta_corrida[k] for k in filas_comparacion}})
comparacion["diferencia"] = comparacion.iloc[:, 1] - comparacion.iloc[:, 0]
display(comparacion.round(4))
print(f"ECE con SAM = {ece:.4f}")

# Brecha train-validación en la mejor época (lo que SAM intenta achicar).
hist_base = pd.read_csv(ruta_baseline / "historial.csv")
brechas = {}
for nombre, h in (("sin SAM", hist_base), ("SAM", historia)):
    fila = h.loc[h["loss_validation"].idxmin()]
    brechas[nombre] = {"mejor_epoca": int(fila["epoca"]), "accuracy_train": fila["accuracy_train"],
                       "accuracy_validation": fila["accuracy_validation"],
                       "brecha_accuracy": fila["accuracy_train"] - fila["accuracy_validation"],
                       "brecha_loss": fila["loss_validation"] - fila["loss_train"]}
display(pd.DataFrame(brechas).T.round(4))

# Bootstrap pareado por panorama.
pred_base = (pd.read_csv(ruta_baseline / "predicciones_validation.csv").set_index("image_id")
             .loc[predicciones["image_id"]])
acierto_sam, acierto_base = predicciones["acierto_top1"].to_numpy(float), pred_base["acierto_top1"].to_numpy(float)
km_sam, km_base = predicciones["error_km"].to_numpy(), pred_base["error_km"].to_numpy()
panoramas, inverso = np.unique(predicciones["panoid"].to_numpy(), return_inverse=True)
filas_por_panorama = [np.flatnonzero(inverso == i) for i in range(len(panoramas))]
rng_boot = np.random.default_rng(SEMILLA)
difs = []
for _ in range(B_BOOTSTRAP):
    ix = np.concatenate([filas_por_panorama[i] for i in rng_boot.integers(0, len(panoramas), len(panoramas))])
    difs.append((acierto_sam[ix].mean() - acierto_base[ix].mean(), np.median(km_sam[ix]) - np.median(km_base[ix])))
difs = np.array(difs)
bootstrap = pd.DataFrame({
    "diferencia": [acierto_sam.mean() - acierto_base.mean(), np.median(km_sam) - np.median(km_base)],
    "ic95_inf": np.percentile(difs, 2.5, axis=0), "ic95_sup": np.percentile(difs, 97.5, axis=0),
}, index=["accuracy_top1 (SAM − sin SAM)", "error_mediano_km (SAM − sin SAM)"])
display(bootstrap.round(4))
print(f"Imágenes que SAM acierta y sin SAM no: {int(((acierto_sam == 1) & (acierto_base == 0)).sum())}; "
      f"al revés: {int(((acierto_sam == 0) & (acierto_base == 1)).sum())}; "
      f"misma celda predicha: {(predicciones['cluster_predicho'].to_numpy() == pred_base['cluster_predicho'].to_numpy()).mean():.3f}")
comparacion.to_csv(ruta_resultados / "comparacion_con_sin_sam.csv")
bootstrap.to_csv(ruta_resultados / "bootstrap_con_sin_sam.csv")

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for ax, columna, titulo in ((axes[0], "accuracy", "Accuracy top-1"), (axes[1], "loss", "Loss haversine")):
    for nombre, h, color in (("sin SAM", hist_base, "tab:blue"), ("SAM", historia, "tab:orange")):
        ax.plot(h["epoca"], h[f"{columna}_train"], "--", color=color, label=f"{nombre} train")
        ax.plot(h["epoca"], h[f"{columna}_validation"], color=color, label=f"{nombre} validación")
    ax.set(xlabel="Época", title=titulo)
    ax.legend()
    ax.grid(alpha=0.3)
fig.tight_layout()
fig.savefig(ruta_resultados / "comparacion_curvas_con_sin_sam.png", dpi=160)
plt.show()
```
