# Formato del dataset
Vamos a ver la cantidad de datos que se tiene el dataset (en la nube)


```python
from datasets import load_dataset

ds = load_dataset(
    "josefbednar/world-streetview-500k",
    split="train",
    streaming=True,
    columns=["country_code"]
)

print(ds)
```

```python
#no correr, 2.5 hs de corrido
from collections import Counter
from tqdm.auto import tqdm

conteo_paises = Counter()

for ejemplo in tqdm(ds):
    conteo_paises[ejemplo["country_code"]] += 1

print("Cantidad de países:", len(conteo_paises))


for pais, cantidad in conteo_paises.most_common():
    print(f"{pais}: {cantidad}")
```

# Busqueda de patornes

```python
# =========================================================
# Leer 6 archivos parquet completos
# =========================================================

import os
import random
import pandas as pd

from PIL import Image
from io import BytesIO

import matplotlib.pyplot as plt


# =========================================================
# CONFIGURACIÓN
# =========================================================

DATA_DIR = r"E:\google_street_view_git\data"

N_ARCHIVOS = 6


# =========================================================
# 1. BUSCAR LOS PARQUETS DISPONIBLES
# =========================================================

archivos = []

for archivo in os.listdir(DATA_DIR):

    if (
        archivo.startswith("train-")
        and archivo.endswith("-of-00393.parquet")
    ):
        archivos.append(archivo)


print(f"Shards encontrados: {len(archivos)}")


# =========================================================
# 2. ELEGIR 6 PARQUETS ALEATORIOS
# =========================================================

if len(archivos) < N_ARCHIVOS:
    raise ValueError(
        f"Solo hay {len(archivos)} archivos disponibles."
    )


archivos_elegidos = random.sample(
    archivos,
    N_ARCHIVOS
)


print("\nArchivos elegidos:")

for archivo in archivos_elegidos:
    print(archivo)


# =========================================================
# 3. LEER TODOS LOS EJEMPLOS DE LOS 6 PARQUETS
# =========================================================

dataframes = []

for archivo in archivos_elegidos:

    ruta = os.path.join(DATA_DIR, archivo)

    print(f"\nLeyendo: {archivo}")

    df_shard = pd.read_parquet(
        ruta,
        columns=[
            "image",
            "country_code",
            "latitude",
            "longitude"
        ]
    )

    print(f"Ejemplos: {len(df_shard)}")

    dataframes.append(df_shard)


# =========================================================
# 4. UNIR LOS 6 DATAFRAMES
# =========================================================

df = pd.concat(dataframes,ignore_index=True)


# =========================================================
# 5. INFORMACIÓN
# =========================================================

print("\n========================================")
print("DATASET")
print("========================================")
print(f"Shape: {df.shape}")
print(f"Cantidad de ejemplos: {len(df)}")
print(f"Columnas: {df.columns.tolist()}")


# =========================================================
# 6. VER UNA IMAGEN
# =========================================================

# Elegimos solamente UNA imagen para decodificar
fila = random.randint(0, len(df) - 1)

imagen_bytes = df.loc[fila, "image"]["bytes"]

imagen = Image.open(
    BytesIO(imagen_bytes)
).convert("RGB")


print("\nImagen seleccionada:")
print(f"Tamaño: {imagen.size}")
print(f"Modo: {imagen.mode}")


plt.figure(figsize=(6, 6))
plt.imshow(imagen)
plt.axis("off")
plt.show()
```

```python
import matplotlib.pyplot as plt

df_plot = df.dropna(
    subset=["latitude", "longitude", "country_code"]
)

# =========================================================
# 1. DISTRIBUCIÓN GEOGRÁFICA
# =========================================================

plt.figure(figsize=(14, 8))
plt.scatter(df_plot["longitude"],df_plot["latitude"],c=df_plot["country_code"].astype("category").cat.codes,s=1,alpha=0.5)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Distribución geográfica de las imágenes")
plt.grid(alpha=0.2)
plt.show()


# =========================================================
# 2. FRECUENCIA POR PAÍS
# =========================================================

frecuencias = df_plot["country_code"].value_counts()

plt.figure(figsize=(12, 10))
plt.barh(frecuencias.index,frecuencias.values)
plt.xlabel("Cantidad de imágenes")
plt.ylabel("País")
plt.title("Frecuencia de imágenes por país")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
