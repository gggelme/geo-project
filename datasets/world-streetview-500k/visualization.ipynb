```python
from datasets import load_dataset

ds = load_dataset(
    "josefbednar/world-streetview-500k",
    streaming=True
)

train = ds["train"]

print(train)
```

```python
train_metadata = train.remove_columns(["image"])
```

```python
from collections import Counter
from tqdm.auto import tqdm

conteo_paises = Counter()

for ejemplo in tqdm(train_metadata):
    conteo_paises[ejemplo["country_code"]] += 1

print("Países distintos:", len(conteo_paises))
```
