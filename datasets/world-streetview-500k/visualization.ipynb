Scripts para ver como trabajar con tal cantidad de datos!!!!


```python
from datasets import load_dataset

ds = load_dataset(
    "josefbednar/world-streetview-500k",
    split="train",
    streaming=True,
    columns=["country_code"]
)

print(ds)
```

```python
#no correr, 2.5 hs de corrido
from collections import Counter
from tqdm.auto import tqdm

conteo_paises = Counter()

for ejemplo in tqdm(ds):
    conteo_paises[ejemplo["country_code"]] += 1

print("Cantidad de países:", len(conteo_paises))


for pais, cantidad in conteo_paises.most_common():
    print(f"{pais}: {cantidad}")
```
